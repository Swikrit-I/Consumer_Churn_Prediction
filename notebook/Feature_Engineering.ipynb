```python
import numpy as np
import pandas as pd
import matplotlib as plt
```

```python
df=pd.read_csv("/Users/swikritsubedi/Desktop/Python/Consumer Churn Prediction/data/Interim/cleaned.csv")
```

Feature Encoding

```python
#Selecting Features for feature encoding
binary_encoding=["gender","Partner","Dependents","PhoneService","PaperlessBilling","Churn"]
one_hot_encoding=["MultipleLines","InternetService","OnlineSecurity","OnlineBackup","DeviceProtection","TechSupport","StreamingTV","StreamingMovies","PaymentMethod"]
ordinal_encoding=["Contract"]
scalerization=["tenure","MonthlyCharges","TotalCharges"]

#ranking order for onehot encoding
contract_order=["Month-to-month","One year","Two year"]
```

```python
#performing binary encoding
maping={"Yes":1,"No":0,"Male":1,"Female":0}
df[binary_encoding]=df[binary_encoding].replace(maping)
```

```python
print(df["TotalCharges"])
```

```python
#Performing Onehot and Ordinal Encoding at once with the help of column transformer
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder,StandardScaler
from sklearn.compose import ColumnTransformer


transformer=ColumnTransformer(
    transformers=[
        ("nomial",OneHotEncoder(drop="first",sparse_output=False),one_hot_encoding),
        ("ordinal",OrdinalEncoder(categories=[contract_order]),ordinal_encoding),
        ("scalerization",StandardScaler(),scalerization)
    ],
    remainder="passthrough",
)
processed_data=transformer.fit_transform(df)
features_name=transformer.get_feature_names_out()
```

```python
df=pd.DataFrame(processed_data,columns=features_name)
df.head()
```
