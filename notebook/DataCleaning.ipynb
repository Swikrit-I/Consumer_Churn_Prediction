```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
#importing raw data file
df=pd.read_csv("/Users/swikritsubedi/Desktop/Python/Consumer Churn Prediction/data/Raw/raw.csv")
```

DataCleaning Starts

```python
#Identificatioon of data which cant be converted to float from string in column total charges
# This filters and shows only the rows that CANNOT be converted to a float
invalid_rows = df[pd.to_numeric(df['TotalCharges'], errors='coerce').isna()]
print(invalid_rows['TotalCharges'].unique())
```

```python
#Replacing empty string with NA
df["TotalCharges"] = df["TotalCharges"].replace(r"^\s*$", np.nan, regex=True)
```

```python
#Forcing The column to be entirely numeric
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
```

```python
#Handeling missing values
imputer=SimpleImputer(strategy="mean")
df["TotalCharges"]=imputer.fit_transform(df[["TotalCharges"]])
```

```python
#Saving The cleaned data
df.to_csv("/Users/swikritsubedi/Desktop/Python/Consumer Churn Prediction/data/Interim/cleaned.csv",index=False)
```
